```python
import pandas as pd
import matplotlib.pyplot as plt


file_path = "Laporan FH Januari 2026.xlsx"


df_raw = pd.read_excel(file_path, sheet_name="FH ", header=4)


df_raw = df_raw.rename(columns={
    df_raw.columns[0]: "No",
    df_raw.columns[1]: "Pos"
})

df_raw.head()
```

```python

date_values = df_raw.iloc[0, 2:]

valid_dates = []
for val in date_values:
    try:
        parsed = pd.to_datetime(val)
        valid_dates.append(parsed)
    except:
        break

n_dates = len(valid_dates)

print("Jumlah periode:", n_dates)
```

```python

df_data = df_raw.iloc[2:].reset_index(drop=True)


asset_row = df_data[df_data["Pos"] == "Total asset"]
dpk_row = df_data[df_data["Pos"] == "Total DPK"]

asset_values = asset_row.iloc[0, 2:2+n_dates].values
dpk_values = dpk_row.iloc[0, 2:2+n_dates].values

ts_df = pd.DataFrame({
    "Date": valid_dates,
    "Total Asset": asset_values,
    "Total DPK": dpk_values
}).sort_values("Date")

ts_df.tail()
```

```python
jan_2026 = ts_df[ts_df["Date"] == "2026-01-01"]
dec_2025 = ts_df[ts_df["Date"] == "2025-12-01"]

asset_growth = jan_2026["Total Asset"].values[0] - dec_2025["Total Asset"].values[0]
dpk_growth = jan_2026["Total DPK"].values[0] - dec_2025["Total DPK"].values[0]

asset_growth_pct = asset_growth / dec_2025["Total Asset"].values[0]
dpk_growth_pct = dpk_growth / dec_2025["Total DPK"].values[0]

print("=== GROWTH JAN 2026 vs DES 2025 ===")
print("Asset Growth (Rp):", format(asset_growth, ","))
print("Asset Growth (%):", round(asset_growth_pct * 100, 2), "%")
print("DPK Growth (Rp):", format(dpk_growth, ","))
print("DPK Growth (%):", round(dpk_growth_pct * 100, 2), "%")
```

SHOCK PERIOD

```python
ts_df["Asset_Return"] = ts_df["Total Asset"].pct_change()
ts_df["DPK_Return"] = ts_df["Total DPK"].pct_change()
```

```python
# Hitung mean dan std
asset_mean = ts_df["Asset_Return"].mean()
asset_std = ts_df["Asset_Return"].std()

dpk_mean = ts_df["DPK_Return"].mean()
dpk_std = ts_df["DPK_Return"].std()

# Z-score
ts_df["Asset_Z"] = (ts_df["Asset_Return"] - asset_mean) / asset_std
ts_df["DPK_Z"] = (ts_df["DPK_Return"] - dpk_mean) / dpk_std
```

```python
# Shock threshold
threshold = 2

asset_shock = ts_df[abs(ts_df["Asset_Z"]) > threshold]
dpk_shock = ts_df[abs(ts_df["DPK_Z"]) > threshold]

print("=== ASSET SHOCK PERIOD ===")
print(asset_shock[["Date", "Asset_Return", "Asset_Z"]])

print("\n=== DPK SHOCK PERIOD ===")
print(dpk_shock[["Date", "DPK_Return", "DPK_Z"]])
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(10,5))
plt.plot(ts_df["Date"], ts_df["Asset_Return"])
plt.scatter(asset_shock["Date"], asset_shock["Asset_Return"])
plt.title("Asset Shock Detection")
plt.xlabel("Date")
plt.ylabel("Return")
plt.xticks(rotation=45)
plt.show()

plt.figure(figsize=(10,5))
plt.plot(ts_df["Date"], ts_df["DPK_Return"])
plt.scatter(dpk_shock["Date"], dpk_shock["DPK_Return"])
plt.title("DPK Shock Detection")
plt.xlabel("Date")
plt.ylabel("Return")
plt.xticks(rotation=45)
plt.show()
```

Terdeteksi ada 2 tipe shock.

Shock positif:
- deposito/tabungan naik
- kredit naik

Shock negatif
- withdraw uang yang banyak

RISK MANAGEMENT

```python
# Hitung perubahan nominal bulanan
ts_df["Asset_Change"] = ts_df["Total Asset"].diff()
ts_df["DPK_Change"] = ts_df["Total DPK"].diff()

# Liquidity gap (DPK minus Asset change)
ts_df["Liquidity_Gap"] = ts_df["DPK_Change"] - ts_df["Asset_Change"]

ts_df[["Date","Liquidity_Gap"]].tail()
```

```python
ts_df["Asset_DPK_Ratio"] = ts_df["Total Asset"] / ts_df["Total DPK"]

ts_df[["Date","Asset_DPK_Ratio"]].tail()
```

```python
# Normalisasi liquidity gap
gap_mean = ts_df["Liquidity_Gap"].mean()
gap_std = ts_df["Liquidity_Gap"].std()

ts_df["Gap_Z"] = (ts_df["Liquidity_Gap"] - gap_mean) / gap_std

# Composite Stress Index
ts_df["Stress_Index"] = (
    abs(ts_df["Asset_Z"]) +
    abs(ts_df["DPK_Z"]) +
    abs(ts_df["Gap_Z"])
)

ts_df[["Date","Stress_Index"]].sort_values("Stress_Index", ascending=False).head()
```

```python
stress_threshold = ts_df["Stress_Index"].mean() + ts_df["Stress_Index"].std()

ews = ts_df[ts_df["Stress_Index"] > stress_threshold]

print("=== EARLY WARNING PERIOD ===")
print(ews[["Date","Stress_Index"]])
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(10,5))
plt.plot(ts_df["Date"], ts_df["Stress_Index"])
plt.axhline(stress_threshold)
plt.title("Composite Stress Index - BPR")
plt.xlabel("Date")
plt.ylabel("Stress Index")
plt.xticks(rotation=45)
plt.show()
```

```python
# Pastikan sudah ada kedua rasio
ts_df["LDR_percent"] = (ts_df["Total Kredit"] / ts_df["Total DPK"]) * 100

ts_df["Modified_LDR_percent"] = (
    ts_df["Total Kredit"] /
    (ts_df["Total DPK"] + ts_df["Total Ekuitas"])
) * 100

# Ambil nilai terakhir
latest = ts_df.iloc[-1]

print("=== PERBANDINGAN RASIO TERAKHIR ===")
print("Tanggal:", latest["Date"])
print("LDR (%):", round(latest["LDR_percent"], 2))
print("LDR + Ekuitas (%):", round(latest["Modified_LDR_percent"], 2))

# Hitung rata-rata periode
print("\n=== RATA-RATA PERIODE ===")
print("Average LDR (%):", round(ts_df["LDR_percent"].mean(), 2))
print("Average LDR + Ekuitas (%):", round(ts_df["Modified_LDR_percent"].mean(), 2))

# Selisih rata-rata
selisih = ts_df["LDR_percent"].mean() - ts_df["Modified_LDR_percent"].mean()

print("\nSelisih rata-rata rasio:", round(selisih, 2), "%")
```

```python
import pandas as pd

# Load tanpa header dulu
df_bopo = pd.read_excel(file_path, sheet_name="BOPO", header=None)

# Ambil bagian atas (2019-2020)
df1 = df_bopo.iloc[0:3].copy()

# Ambil bagian bawah (2023)
df2 = df_bopo.iloc[3:6].copy()

# Gabungkan keduanya
df_bopo_clean = pd.concat([df1, df2])

# Jadikan baris pertama sebagai header
df_bopo_clean.columns = df_bopo_clean.iloc[0]
df_bopo_clean = df_bopo_clean[1:]

df_bopo_clean.rename(columns={df_bopo_clean.columns[0]: "Pos"}, inplace=True)

df_bopo_clean.head()
```

```python
df_bopo_long = df_bopo_clean.melt(id_vars="Pos",
                                  var_name="Date",
                                  value_name="Value")

df_bopo_long.head()
```

```python
df_bopo_long["Value"] = (
    df_bopo_long["Value"]
    .astype(str)
    .str.replace("Rp", "", regex=False)
    .str.replace(",", "", regex=False)
    .str.strip()
)

df_bopo_long["Value"] = pd.to_numeric(df_bopo_long["Value"], errors="coerce")
```

```python
df_bopo_long["Date"] = pd.to_datetime(df_bopo_long["Date"], errors="coerce")
```

```python
import pandas as pd

file_path = "ii02.XLSX"

# Baca dua sheet
df_old = pd.read_excel(file_path, sheet_name="ii02", header=None)
df_new = pd.read_excel(file_path, sheet_name="ii02_BARU", header=None)

def extract_bpr(df_raw):

    # Cari baris BPR
    bpr_idx = df_raw[df_raw[1].astype(str)
                     .str.contains("Bank Perkreditan Rakyat", na=False)].index[0]

    tab_row = df_raw.iloc[bpr_idx + 1]
    dep_row = df_raw.iloc[bpr_idx + 3]

    # Cari baris tahun (biasanya di row ke-4)
    years = df_raw.iloc[4]
    year_cols = years[years.apply(lambda x: str(x).isdigit())].index

    tab = pd.to_numeric(tab_row[year_cols], errors="coerce")
    dep = pd.to_numeric(dep_row[year_cols], errors="coerce")

    ts = pd.DataFrame({
        "Year": years[year_cols].values,
        "Tabungan_BI": tab.values,
        "Deposito_BI": dep.values
    })

    return ts


# Extract masing-masing sheet
ts_old = extract_bpr(df_old)
ts_new = extract_bpr(df_new)

# Gabungkan
ts_all = pd.concat([ts_old, ts_new], ignore_index=True)

# Bersihkan
ts_all = ts_all.dropna()
ts_all["Year"] = ts_all["Year"].astype(int)

# Kalau ada duplikat tahun (misalnya 2024 muncul 2x)
ts_all = ts_all.groupby("Year", as_index=False).mean()

ts_all = ts_all.sort_values("Year").reset_index(drop=True)

ts_all
```

```python
import pandas as pd

file_path = "ii02.XLSX"

# Load both sheets again to ensure df_old and df_new are available in this scope
df_old = pd.read_excel(file_path, sheet_name="ii02", header=None)
df_new = pd.read_excel(file_path, sheet_name="ii02_BARU", header=None)

def extract_bpr_kredit(df_raw):
    # Find the row containing "Bank Perkreditan Rakyat"
    bpr_idx_list = df_raw[df_raw[1].astype(str).str.contains("Bank Perkreditan Rakyat", na=False)].index.tolist()
    if not bpr_idx_list:
        return pd.DataFrame() # Return empty if not found

    bpr_idx = bpr_idx_list[0]

    # Based on inspection of df_raw.head(40) in earlier cells,
    # "Kredit yang diberikan" is typically at bpr_idx + 4
    kredit_row = df_raw.iloc[bpr_idx + 4]

    # Find the year rows (assuming years are consistently in row 4)
    years = df_raw.iloc[4]
    year_cols = years[years.apply(lambda x: str(x).isdigit())].index

    kredit_values = pd.to_numeric(kredit_row[year_cols], errors="coerce")

    ts = pd.DataFrame({
        "Year": years[year_cols].values,
        "Kredit_BI": kredit_values.values
    })
    return ts

# Extract credit data from both sheets
bi_kredit_old = extract_bpr_kredit(df_old)
bi_kredit_new = extract_bpr_kredit(df_new)

# Concatenate them
bi_kredit_df = pd.concat([bi_kredit_old, bi_kredit_new], ignore_index=True)

# Clean and aggregate by year (handling potential duplicates from multiple sheets)
bi_kredit_df = bi_kredit_df.dropna(subset=['Kredit_BI'])
bi_kredit_df["Year"] = bi_kredit_df["Year"].astype(int)
bi_kredit_df = bi_kredit_df.groupby("Year", as_index=False).mean()
bi_kredit_df = bi_kredit_df.sort_values("Year").reset_index(drop=True)

# Pastikan hanya sampai 2024
bi_kredit_df = bi_kredit_df[bi_kredit_df["Year"] <= 2024]

# Urutkan
bi_kredit_df = bi_kredit_df.sort_values("Year")

# Konversi satuan juta → rupiah
bi_kredit_df["Kredit_BI_Rp"] = bi_kredit_df["Kredit_BI"] * 1_000_000

# Hitung YoY
bi_kredit_df["Kredit_YoY_BI_%"] = (
    bi_kredit_df["Kredit_BI_Rp"].pct_change() * 100
)
```
